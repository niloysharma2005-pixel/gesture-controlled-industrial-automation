```python
# =============================================================================
# HGAG FINAL SUBMISSION NOTEBOOK -- KAGGLE VERSION (GPU)
# =============================================================================
# Paste this ENTIRE file into a SINGLE Kaggle notebook cell and run.
#
# This final notebook is structured to satisfy the academic project rubric while
# preserving the leakage-safe, post-audit evaluation design.
#
# Included modules:
#   1. Problem Understanding / Dataset Analysis
#   2. Audit-backed Preprocessing
#   3. Benchmark Baseline (handcrafted features + Logistic Regression)
#   4. CNN-LSTM LOSO Training
#   5. Evaluation / Visualization
#   6. XAI (1D Grad-CAM on representative fold)
#   7. Deployment / Industrial Readiness
#   8. Model Optimization (TFLite dynamic quantization)
#   9. Creative Dashboard + Final Documentation Exports
# =============================================================================

import gc
import io
import json
import os
import traceback
import time

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import tensorflow as tf
from scipy.optimize import linear_sum_assignment
from scipy.signal import butter, filtfilt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_recall_fscore_support,
    roc_auc_score,
)
from sklearn.preprocessing import StandardScaler, label_binarize
from tensorflow.keras import callbacks as cb
from tensorflow.keras import layers, models


RUN_INPUT_PATH_SCAN = False
RUN_BASELINE_BENCHMARK = True
RUN_XAI = True
RUN_MODEL_OPTIMIZATION = True
RUN_STREAMING_DEMO = True


if RUN_INPUT_PATH_SCAN:
    for dirname, _, filenames in os.walk("/kaggle/input"):
        for filename in filenames[:5]:
            print(os.path.join(dirname, filename))
        if filenames:
            print(f"  ... ({len(filenames)} files total in {dirname})")


# -----------------------------------------------------------------------------
# Paths
# -----------------------------------------------------------------------------
RAW_PATH = (
    "/kaggle/input/datasets/hlfbldprnc/hgag-data/"
    "Hand Gesture Accelerometer and Gyroscope Dataset (HGAG-DATA)/"
    "HGAG-DATA/HGAG-DATA1"
)
OUTPUT_DIR = "/kaggle/working/hgag_clean_folds"
RESULTS_DIR = "/kaggle/working/hgag_results_final"


# -----------------------------------------------------------------------------
# Project defaults locked from the verified preprocessing audit
# -----------------------------------------------------------------------------
N_SUBJECTS = 43
SR = 200

WINDOW_OFFSET = 50
WIN_SIZE = 200
STEP_SIZE = 200

PREPROCESS_MODE = "released"
VAL_SUBJECTS_PER_FOLD = 4

EPOCHS = 25
BATCH = 128
PATIENCE = 7
CONFIDENCE_GATE = 0.70

GESTURE_FOLDERS = [
    "Clapping",
    "Coin Flipping",
    "Finger Snapping",
    "Fist Making",
    "Horizontal Wrist Extension",
    "Index Finger Flick",
    "Index Thumb Tap",
    "Shooting",
    "Thumb Up",
    "Wrist Extension",
    "Wrist Flexion",
]
CHANNEL_FILES = [
    "accel_x_data.csv",
    "accel_y_data.csv",
    "accel_z_data.csv",
    "gyro_x_data.csv",
    "gyro_y_data.csv",
    "gyro_z_data.csv",
]
GESTURE_NAMES = [
    "Clapping",
    "Coin Flip",
    "Finger Snap",
    "Fist Making",
    "Horiz. Wrist Ext",
    "Idx Finger Flick",
    "Idx Thumb Tap",
    "Shooting",
    "Thumb Up",
    "Wrist Ext",
    "Wrist Flex",
]
CHANNEL_NAMES = ["acc_x", "acc_y", "acc_z", "gyro_x", "gyro_y", "gyro_z"]

N_GESTURES = len(GESTURE_FOLDERS)
N_CHANNELS = len(CHANNEL_FILES)

AUG_CONFIG = {
    "jitter_sigma": 0.03,
    "scale_accel_range": (0.85, 1.15),
    "scale_gyro_range": (0.90, 1.10),
    "warp_fraction": 0.25,
    "warp_factor_range": (0.8, 1.2),
}

ARCH_RATIONALE = """
Architecture: CNN-LSTM for IMU gesture recognition

Conv1D block (x3, kernels 5-5-3, filters 64-128-128):
  Extracts local temporal motions before recurrent modelling. Kernel size 5
  corresponds to about 25 ms at 200 Hz, which is suitable for short gesture
  transients and wrist-motion bursts. The later kernel size 3 layer refines
  finer temporal details after broader local patterns have already been formed.

MaxPooling1D (x2):
  Compresses the temporal axis from 200 -> 100 -> 50 steps before the LSTM.
  This reduces recurrent cost and noise sensitivity while preserving enough
  motion structure for sequence modelling.

BatchNormalization:
  Stabilises training across subjects whose motion amplitudes differ even after
  preprocessing and train-only normalization.

LSTM (64 hidden units):
  Models inter-frame dependencies that local convolutions cannot capture alone.
  64 hidden units provide enough sequence capacity without becoming too large
  for a 43-subject LOSO setting where overfitting is a real risk.

Dropout (0.3 / 0.4 / 0.3):
  Uses conservative regularisation, with slightly stronger dropout around the
  recurrent and dense stages to reduce subject memorisation.

Adam + EarlyStopping + ReduceLROnPlateau:
  Adam provides robust optimisation for this medium-scale time-series task.
  LR reduction smooths late convergence, while early stopping prevents
  unnecessary fit to fold-specific validation dynamics.
"""

COMMANDS = [
    {"name": "EMERGENCY STOP", "criticality": 10},
    {"name": "Start Machine", "criticality": 8},
    {"name": "Move Arm Left", "criticality": 6},
    {"name": "Move Arm Right", "criticality": 6},
    {"name": "Activate Tool", "criticality": 7},
    {"name": "Deactivate Tool", "criticality": 7},
    {"name": "Speed Up", "criticality": 5},
    {"name": "Slow Down", "criticality": 5},
    {"name": "Confirm Action", "criticality": 4},
    {"name": "Cancel Action", "criticality": 4},
    {"name": "Call Supervisor", "criticality": 3},
]


def print_banner(title):
    print("\n" + "=" * 78)
    print(f"  {title}")
    print("=" * 78)


def save_json(path, payload):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(payload, f, indent=2)


def ensure_supported_preprocess(mode):
    valid = {"released", "raw", "bp30_90", "lp90"}
    if mode not in valid:
        raise ValueError(f"Unsupported PREPROCESS_MODE '{mode}'. Valid modes: {sorted(valid)}")


ensure_supported_preprocess(PREPROCESS_MODE)


def clean(signal, mode=None):
    """Apply the selected signal preprocessing to one (time, 6) trial."""
    mode = PREPROCESS_MODE if mode is None else mode
    mode = "released" if mode == "raw" else mode

    if mode == "released":
        return signal.astype(np.float32)

    nyq = SR / 2.0

    if mode == "bp30_90":
        b, a = butter(4, [30.0 / nyq, 90.0 / nyq], btype="bandpass")
        return filtfilt(b, a, signal, axis=0).astype(np.float32)

    if mode == "lp90":
        b, a = butter(4, 90.0 / nyq, btype="low")
        return filtfilt(b, a, signal, axis=0).astype(np.float32)

    raise ValueError(f"Unknown preprocess mode: {mode}")


def apply_audit_filter_mode(signal, mode):
    """Apply audit comparison filters for runtime evidence generation."""
    if mode == "released":
        return signal.astype(np.float32)

    nyq = SR / 2.0

    if mode == "lp90":
        b, a = butter(4, 90.0 / nyq, btype="low")
        return filtfilt(b, a, signal, axis=0).astype(np.float32)

    if mode == "bp30_90":
        b, a = butter(4, [30.0 / nyq, 90.0 / nyq], btype="bandpass")
        return filtfilt(b, a, signal, axis=0).astype(np.float32)

    if mode == "lp40_accel_detrend":
        b, a = butter(4, 40.0 / nyq, btype="low")
        x = filtfilt(b, a, signal, axis=0).astype(np.float32)
        b2, a2 = butter(2, 0.6 / nyq, btype="low")
        for c in range(3):
            gravity = filtfilt(b2, a2, x[:, c])
            x[:, c] = x[:, c] - gravity
        return x

    raise ValueError(f"Unknown audit mode: {mode}")


def segment(signal):
    """Segment one preprocessed trial into post-trim windows."""
    n = len(signal)
    starts = list(range(WINDOW_OFFSET, n - WIN_SIZE + 1, STEP_SIZE))
    if len(starts) == 0:
        return np.empty((0, WIN_SIZE, N_CHANNELS), dtype=np.float32)
    return np.stack([signal[s : s + WIN_SIZE] for s in starts]).astype(np.float32)


def load_trial_signal(subject_id, folder, rep_idx=0):
    """Load one raw HGAG repetition from disk."""
    csv_dir = os.path.join(RAW_PATH, folder, f"Subject_{subject_id}", ".csv")
    if not os.path.isdir(csv_dir):
        return None

    ch_arrays = []
    for fname in CHANNEL_FILES:
        fp = os.path.join(csv_dir, fname)
        if not os.path.exists(fp):
            return None
        ch_arrays.append(pd.read_csv(fp, header=None).values.astype(np.float32))

    n_reps = min(arr.shape[0] for arr in ch_arrays)
    if rep_idx >= n_reps:
        return None

    return np.stack([arr[rep_idx] for arr in ch_arrays], axis=1).astype(np.float32)


def iter_raw_trials():
    """Yield every raw trial from the dataset for runtime audit computation."""
    for sid in range(1, N_SUBJECTS + 1):
        for gesture_idx, folder in enumerate(GESTURE_FOLDERS):
            csv_dir = os.path.join(RAW_PATH, folder, f"Subject_{sid}", ".csv")
            if not os.path.isdir(csv_dir):
                continue

            ch_arrays = []
            ok = True
            for fname in CHANNEL_FILES:
                fp = os.path.join(csv_dir, fname)
                if not os.path.exists(fp):
                    ok = False
                    break
                ch_arrays.append(pd.read_csv(fp, header=None).values.astype(np.float32))
            if not ok:
                continue

            n_reps = min(arr.shape[0] for arr in ch_arrays)
            for rep_idx in range(n_reps):
                raw_sig = np.stack([arr[rep_idx] for arr in ch_arrays], axis=1).astype(np.float32)
                yield sid, gesture_idx, rep_idx, raw_sig


def compute_runtime_audit_summary(results_dir):
    """Compute audit evidence directly from the dataset during the final run."""
    print_banner("RUNTIME AUDIT SUMMARY")

    audit_modes = {
        "released": "Released signal",
        "lp90": "90 Hz low-pass",
        "bp30_90": "30-90 Hz band-pass",
        "lp40_accel_detrend": "40 Hz low-pass + accel detrend",
    }

    trial_lengths = []
    quiet_ratios = {name: [] for name in CHANNEL_NAMES}
    retention = {mode: {name: [] for name in CHANNEL_NAMES} for mode in audit_modes}
    subjects_seen = set()
    trial_count = 0

    for sid, _gesture_idx, _rep_idx, raw_sig in iter_raw_trials():
        subjects_seen.add(sid)
        trial_count += 1
        trial_lengths.append(int(raw_sig.shape[0]))

        k = min(WINDOW_OFFSET, raw_sig.shape[0] - 1)
        lead = np.abs(raw_sig[:k])
        tail = np.abs(raw_sig[k:])
        lead_mean = np.mean(lead, axis=0)
        tail_mean = np.mean(tail, axis=0)
        ratio = lead_mean / np.maximum(tail_mean, 1e-8)
        for ch_idx, ch_name in enumerate(CHANNEL_NAMES):
            quiet_ratios[ch_name].append(float(ratio[ch_idx]))

        raw_dyn = raw_sig - np.mean(raw_sig, axis=0, keepdims=True)
        raw_rms = np.sqrt(np.mean(raw_dyn**2, axis=0))

        for mode in audit_modes:
            proc = apply_audit_filter_mode(raw_sig.copy(), mode)
            proc_dyn = proc - np.mean(proc, axis=0, keepdims=True)
            proc_rms = np.sqrt(np.mean(proc_dyn**2, axis=0))
            rms_ret = proc_rms / np.maximum(raw_rms, 1e-8) * 100.0
            for ch_idx, ch_name in enumerate(CHANNEL_NAMES):
                retention[mode][ch_name].append(float(rms_ret[ch_idx]))

    if trial_count == 0:
        raise RuntimeError("Runtime audit failed: no raw trials were found.")

    summary = {
        "source": "computed_during_final_notebook_run",
        "sampling_rate_hz": SR,
        "subjects_audited": sorted(subjects_seen),
        "n_trials_audited": trial_count,
        "trial_length_samples": int(round(float(np.median(trial_lengths)))),
        "trial_length_ms": int(round(float(np.median(trial_lengths)) / SR * 1000.0)),
        "current_offset_reference_ms": int(round(WINDOW_OFFSET / SR * 1000.0)),
        "quiet_leadin_ratio_median_by_channel": {
            ch: float(np.median(vals)) for ch, vals in quiet_ratios.items()
        },
        "rms_retention_pct_by_mode_and_channel": {
            mode: {
                "label": label,
                "channels": {
                    ch: round(float(np.mean(vals)), 2) for ch, vals in retention[mode].items()
                },
            }
            for mode, label in audit_modes.items()
        },
        "suggested_preprocessing_candidates": [
            {"code": "released", "label": "Released signal"},
            {"code": "bp30_90", "label": "30-90 Hz band-pass"},
            {"code": "lp90", "label": "90 Hz low-pass"},
        ],
        "rejected_preprocessing_modes": [
            {
                "code": "lp40_accel_detrend",
                "label": "40 Hz low-pass + accel detrend",
                "reason": (
                    "Runtime audit shows severe signal suppression compared with released data."
                ),
            }
        ],
    }

    save_json(os.path.join(results_dir, "runtime_audit_summary.json"), summary)
    print(
        f"  Audited {trial_count:,} trials | median quiet lead-in ratio "
        f"(gyro_x)={summary['quiet_leadin_ratio_median_by_channel']['gyro_x']:.4f}"
    )
    return summary


def load_subject(sid):
    """Load, preprocess, and segment every repetition for one subject."""
    x_all, y_all = [], []
    for label, folder in enumerate(GESTURE_FOLDERS):
        csv_dir = os.path.join(RAW_PATH, folder, f"Subject_{sid}", ".csv")
        if not os.path.isdir(csv_dir):
            continue

        ch_arrays = []
        ok = True
        for fname in CHANNEL_FILES:
            fp = os.path.join(csv_dir, fname)
            if not os.path.exists(fp):
                ok = False
                break
            ch_arrays.append(pd.read_csv(fp, header=None).values.astype(np.float32))
        if not ok:
            continue

        n_reps = min(arr.shape[0] for arr in ch_arrays)
        for rep in range(n_reps):
            rep_signal = np.stack([arr[rep] for arr in ch_arrays], axis=1)
            windows = segment(clean(rep_signal))
            if len(windows):
                x_all.extend(windows)
                y_all.extend([label] * len(windows))

    if not x_all:
        return None, None

    return np.array(x_all, dtype=np.float32), np.array(y_all, dtype=np.int32)


def choose_validation_subjects(train_sids, fold_id):
    """Deterministic subject-pure validation split inside each LOSO fold."""
    n_val = min(VAL_SUBJECTS_PER_FOLD, max(1, len(train_sids) - 1))
    split_rng = np.random.default_rng(1000 + fold_id)
    shuffled = split_rng.permutation(np.array(sorted(train_sids), dtype=np.int32)).tolist()
    val_sids = sorted(int(s) for s in shuffled[:n_val])
    inner_train_sids = sorted(int(s) for s in shuffled[n_val:])
    return inner_train_sids, val_sids


def _window_warp_batch(x, rng, warp_fraction, warp_factor_range):
    """Apply conservative local time warping to a batch of windows."""
    n, t, c = x.shape
    warp_len = min(t - 2, max(2, int(t * warp_fraction)))
    x_ww = np.empty_like(x)

    for i in range(n):
        start = int(rng.integers(0, t - warp_len + 1))
        end = start + warp_len
        factor = float(rng.uniform(*warp_factor_range))

        seg_before = x[i, :start, :]
        seg_mid = x[i, start:end, :]
        seg_after = x[i, end:, :]

        new_len = max(2, int(round(warp_len * factor)))
        orig_idx = np.linspace(0, warp_len - 1, warp_len)
        new_idx = np.linspace(0, warp_len - 1, new_len)
        seg_mid_warped = np.stack(
            [np.interp(new_idx, orig_idx, seg_mid[:, ch]) for ch in range(c)],
            axis=1,
        )

        stitched = np.concatenate([seg_before, seg_mid_warped, seg_after], axis=0)
        stitched_len = stitched.shape[0]

        if stitched_len == t:
            x_ww[i] = stitched.astype(np.float32)
        else:
            orig_t = np.linspace(0, 1, stitched_len)
            resamp_t = np.linspace(0, 1, t)
            x_ww[i] = np.stack(
                [np.interp(resamp_t, orig_t, stitched[:, ch]) for ch in range(c)],
                axis=1,
            ).astype(np.float32)

    return x_ww.astype(np.float32)


def augment(x, y, rng):
    """Apply train-only augmentation to normalized training windows."""
    n = len(x)

    jittered = (x + rng.normal(0, AUG_CONFIG["jitter_sigma"], x.shape)).astype(np.float32)

    scale = np.ones((n, 1, N_CHANNELS), dtype=np.float32)
    scale[:, :, :3] = rng.uniform(
        *AUG_CONFIG["scale_accel_range"], (n, 1, 1)
    ).astype(np.float32)
    scale[:, :, 3:] = rng.uniform(
        *AUG_CONFIG["scale_gyro_range"], (n, 1, 1)
    ).astype(np.float32)
    scaled = (x * scale).astype(np.float32)

    warped = _window_warp_batch(
        x,
        rng,
        warp_fraction=AUG_CONFIG["warp_fraction"],
        warp_factor_range=AUG_CONFIG["warp_factor_range"],
    )

    parts = [x, jittered, scaled, warped]
    x_aug = np.concatenate(parts, axis=0)
    y_aug = np.tile(y, len(parts))
    idx = rng.permutation(len(x_aug))
    return x_aug[idx], y_aug[idx]


def extract_handcrafted_features(x):
    """Compute compact handcrafted features for a feature-based benchmark."""
    mean = np.mean(x, axis=1)
    std = np.std(x, axis=1)
    min_v = np.min(x, axis=1)
    max_v = np.max(x, axis=1)
    rms = np.sqrt(np.mean(x**2, axis=1))
    abs_mean = np.mean(np.abs(x), axis=1)
    q25 = np.percentile(x, 25, axis=1)
    q75 = np.percentile(x, 75, axis=1)
    ptp = np.ptp(x, axis=1)

    feat = np.concatenate(
        [mean, std, min_v, max_v, rms, abs_mean, q25, q75, ptp],
        axis=1,
    )
    return feat.astype(np.float32)


def plot_dataset_overview(all_data, results_dir):
    """Create dataset coverage and balance plots for the report."""
    subject_rows = []
    gesture_counts = np.zeros(N_GESTURES, dtype=np.int32)
    for sid, (x_sub, y_sub) in all_data.items():
        subject_rows.append(
            {
                "subject": sid,
                "n_windows": len(x_sub),
                "n_classes": len(np.unique(y_sub)),
            }
        )
        binc = np.bincount(y_sub, minlength=N_GESTURES)
        gesture_counts += binc

    subject_df = pd.DataFrame(subject_rows).sort_values("subject")
    subject_df.to_csv(os.path.join(results_dir, "subject_window_counts.csv"), index=False)

    fig, axes = plt.subplots(1, 2, figsize=(16, 5))

    axes[0].bar(subject_df["subject"], subject_df["n_windows"], color="#4c78a8", edgecolor="black", linewidth=0.4)
    axes[0].set_title("Window Count per Subject", fontweight="bold")
    axes[0].set_xlabel("Subject ID")
    axes[0].set_ylabel("Windows")
    axes[0].set_xticks(subject_df["subject"])
    axes[0].grid(axis="y", alpha=0.3)

    axes[1].bar(GESTURE_NAMES, gesture_counts, color="#f58518", edgecolor="black", linewidth=0.4)
    axes[1].set_title("Window Count per Gesture", fontweight="bold")
    axes[1].set_ylabel("Windows")
    axes[1].tick_params(axis="x", rotation=45)
    axes[1].grid(axis="y", alpha=0.3)

    plt.tight_layout()
    out = os.path.join(results_dir, "eda_dataset_overview.png")
    plt.savefig(out, dpi=150)
    plt.close(fig)
    print(f"  Dataset overview -> {out}")


def report_class_balance(all_data, results_dir):
    """Report overall gesture balance after post-audit segmentation."""
    counts = np.zeros(N_GESTURES, dtype=np.int64)
    subject_rows = []
    for sid, (x_sub, y_sub) in all_data.items():
        counts += np.bincount(y_sub, minlength=N_GESTURES)
        subject_rows.append({"subject": sid, "windows": len(x_sub)})

    total = int(counts.sum())
    cv = float(np.std(counts) / max(np.mean(counts), 1e-8))
    gini_num = np.sum(np.abs(counts[:, None] - counts[None, :]))
    gini = float(gini_num / max(2 * N_GESTURES * total, 1e-8))

    print_banner("CLASS BALANCE REPORT")
    print(f"  {'Gesture':<22} {'Windows':>8}  {'% share':>8}")
    print(f"  {'-' * 44}")
    mean_count = total / max(N_GESTURES, 1)
    for gesture, count in zip(GESTURE_NAMES, counts):
        flag = " <-- LOW" if count < 0.7 * mean_count else ""
        print(f"  {gesture:<22} {count:>8,}  {count / total * 100:>7.1f}%{flag}")
    print(f"\n  Coefficient of variation : {cv:.3f}")
    print(f"  Gini imbalance index     : {gini:.4f}  (0=perfect, 1=max)")

    pd.DataFrame(
        {
            "gesture": GESTURE_NAMES,
            "windows": counts,
            "pct_share": counts / total * 100,
        }
    ).to_csv(os.path.join(results_dir, "class_balance_report.csv"), index=False)
    pd.DataFrame(subject_rows).to_csv(os.path.join(results_dir, "subject_window_totals.csv"), index=False)

    return {
        "counts": counts.tolist(),
        "coefficient_of_variation": cv,
        "gini_index": gini,
    }


def plot_raw_window_visuals(all_data, results_dir):
    """Create post-audit window EDA figures."""
    all_x = np.concatenate([xy[0] for xy in all_data.values()], axis=0)
    all_y = np.concatenate([xy[1] for xy in all_data.values()], axis=0)

    flat = all_x.reshape(-1, N_CHANNELS)
    corr = np.corrcoef(flat.T)
    fig, ax = plt.subplots(figsize=(8, 6))
    sns.heatmap(
        corr,
        annot=True,
        fmt=".2f",
        cmap="RdBu_r",
        vmin=-1,
        vmax=1,
        xticklabels=CHANNEL_NAMES,
        yticklabels=CHANNEL_NAMES,
        ax=ax,
        linewidths=0.5,
        square=True,
    )
    ax.set_title(f"Channel Correlation ({PREPROCESS_MODE} windows)", fontsize=13, fontweight="bold")
    plt.tight_layout()
    out = os.path.join(results_dir, "eda_correlation.png")
    plt.savefig(out, dpi=150)
    plt.close(fig)
    print(f"  Correlation figure -> {out}")

    fig, axes = plt.subplots(3, 4, figsize=(20, 12))
    axes = axes.flatten()
    for g in range(N_GESTURES):
        ax = axes[g]
        mask = all_y == g
        if np.any(mask):
            mean_sig = all_x[mask].mean(axis=0)
            for ch in range(3):
                ax.plot(mean_sig[:, ch], label=CHANNEL_NAMES[ch], linewidth=1.5)
            for ch in range(3, 6):
                ax.plot(mean_sig[:, ch], label=CHANNEL_NAMES[ch], linewidth=1.0, linestyle="--")
        ax.set_title(GESTURE_NAMES[g], fontsize=10, fontweight="bold")
        ax.grid(True, alpha=0.2)
        if g == 0:
            ax.legend(fontsize=7)
    axes[11].axis("off")
    plt.suptitle(
        f"Mean Gesture Window Profiles ({PREPROCESS_MODE} preprocessing)",
        fontsize=14,
        fontweight="bold",
    )
    plt.tight_layout()
    out = os.path.join(results_dir, "eda_profiles.png")
    plt.savefig(out, dpi=150)
    plt.close(fig)
    print(f"  Gesture-profile figure -> {out}")

    energy = np.zeros((N_GESTURES, N_CHANNELS))
    for g in range(N_GESTURES):
        mask = all_y == g
        if np.any(mask):
            energy[g] = np.sqrt(np.mean(all_x[mask] ** 2, axis=(0, 1)))
    energy_norm = energy / np.maximum(energy.max(axis=0, keepdims=True), 1e-8)
    fig, ax = plt.subplots(figsize=(12, 6))
    sns.heatmap(
        energy_norm,
        annot=True,
        fmt=".2f",
        cmap="YlOrRd",
        xticklabels=CHANNEL_NAMES,
        yticklabels=GESTURE_NAMES,
        ax=ax,
        linewidths=0.5,
    )
    ax.set_title(f"Channel Activation Energy ({PREPROCESS_MODE} windows)", fontsize=13, fontweight="bold")
    plt.tight_layout()
    out = os.path.join(results_dir, "eda_energy.png")
    plt.savefig(out, dpi=150)
    plt.close(fig)
    print(f"  Energy figure -> {out}")


def plot_released_signal_example(results_dir):
    """Save one representative released-signal example."""
    ref_subject = 1
    ref_gesture_idx = 0
    raw_trial = load_trial_signal(ref_subject, GESTURE_FOLDERS[ref_gesture_idx], 0)

    if raw_trial is None:
        for sid in range(1, N_SUBJECTS + 1):
            for gesture_idx, folder in enumerate(GESTURE_FOLDERS):
                raw_trial = load_trial_signal(sid, folder, 0)
                if raw_trial is not None:
                    ref_subject = sid
                    ref_gesture_idx = gesture_idx
                    break
            if raw_trial is not None:
                break

    if raw_trial is None:
        print("  Released signal example skipped: no trial found.")
        return

    t_axis = np.arange(raw_trial.shape[0]) / SR * 1000.0
    fig, axes = plt.subplots(3, 2, figsize=(14, 8), sharex=True)
    axes = axes.flatten()
    for ch, ax in enumerate(axes):
        ax.plot(t_axis, raw_trial[:, ch], linewidth=1.3)
        ax.axvspan(0, WINDOW_OFFSET / SR * 1000.0, alpha=0.12, color="orange")
        ax.set_title(CHANNEL_NAMES[ch], fontweight="bold")
        ax.set_ylabel("Amplitude")
        ax.grid(True, alpha=0.2)
    axes[-2].set_xlabel("Time (ms)")
    axes[-1].set_xlabel("Time (ms)")
    plt.suptitle(
        f"Released Signal Example | {GESTURE_NAMES[ref_gesture_idx]} | Subject {ref_subject}",
        fontsize=13,
        fontweight="bold",
    )
    plt.tight_layout()
    out = os.path.join(results_dir, "eda_released_signal_example.png")
    plt.savefig(out, dpi=150, bbox_inches="tight")
    plt.close(fig)
    print(f"  Released example -> {out}")


def plot_augmentation_preview(sample, results_dir):
    """Visualize the training augmentations on one normalized window."""
    preview_rng = np.random.default_rng(2026)
    sample_batch = sample[np.newaxis, ...]

    jittered = (
        sample_batch + preview_rng.normal(0, AUG_CONFIG["jitter_sigma"], sample_batch.shape)
    ).astype(np.float32)[0]

    scale = np.ones((1, 1, N_CHANNELS), dtype=np.float32)
    scale[:, :, :3] = preview_rng.uniform(
        *AUG_CONFIG["scale_accel_range"], (1, 1, 1)
    ).astype(np.float32)
    scale[:, :, 3:] = preview_rng.uniform(
        *AUG_CONFIG["scale_gyro_range"], (1, 1, 1)
    ).astype(np.float32)
    scaled = (sample_batch * scale).astype(np.float32)[0]

    warped = _window_warp_batch(
        sample_batch,
        preview_rng,
        warp_fraction=AUG_CONFIG["warp_fraction"],
        warp_factor_range=AUG_CONFIG["warp_factor_range"],
    )[0]

    variants = [
        ("Original (normalized)", sample),
        ("Jittered", jittered),
        ("Scaled", scaled),
        ("Window-warped", warped),
    ]

    fig, axes = plt.subplots(2, 2, figsize=(15, 8), sharex=True)
    axes = axes.flatten()
    for ax, (title, arr) in zip(axes, variants):
        for ch in range(3):
            ax.plot(arr[:, ch], linewidth=1.5, label=CHANNEL_NAMES[ch] if title == "Original (normalized)" else None)
        for ch in range(3, 6):
            ax.plot(arr[:, ch], linewidth=1.0, linestyle="--")
        ax.set_title(title, fontweight="bold")
        ax.grid(True, alpha=0.2)
    axes[0].legend(fontsize=8)
    plt.suptitle("Augmentation Preview on One Training Window", fontsize=14, fontweight="bold")
    plt.tight_layout()
    out = os.path.join(results_dir, "augmentation_preview.png")
    plt.savefig(out, dpi=150, bbox_inches="tight")
    plt.close(fig)
    print(f"  Augmentation preview -> {out}")


def plot_training_curves(history, fold_id, results_dir):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    axes[0].plot(history.history["loss"], label="Train")
    axes[0].plot(history.history["val_loss"], label="Val")
    axes[0].set_title(f"Fold {fold_id:02d} Loss", fontweight="bold")
    axes[0].set_xlabel("Epoch")
    axes[0].set_ylabel("Loss")
    axes[0].grid(True, alpha=0.25)
    axes[0].legend()

    axes[1].plot(history.history["accuracy"], label="Train")
    axes[1].plot(history.history["val_accuracy"], label="Val")
    axes[1].set_title(f"Fold {fold_id:02d} Accuracy", fontweight="bold")
    axes[1].set_xlabel("Epoch")
    axes[1].set_ylabel("Accuracy")
    axes[1].grid(True, alpha=0.25)
    axes[1].legend()

    plt.tight_layout()
    out = os.path.join(results_dir, f"training_curves_fold_{fold_id:02d}.png")
    plt.savefig(out, dpi=150, bbox_inches="tight")
    plt.close(fig)
    print(f"  Training curves -> {out}")


def plot_all_fold_curves(all_histories, results_dir):
    """Plot mean +/- std convergence across all LOSO folds."""
    if not all_histories:
        return

    max_ep = max(len(h["val_loss"]) for h in all_histories)

    def pad(arr):
        return arr + [np.nan] * (max_ep - len(arr))

    keys = ["loss", "val_loss", "accuracy", "val_accuracy"]
    mats = {k: np.array([pad(h[k]) for h in all_histories], dtype=np.float32) for k in keys}
    ep = np.arange(1, max_ep + 1)

    fig, axes = plt.subplots(1, 2, figsize=(14, 4))
    for ax, (tr, vl, ylabel) in zip(
        axes,
        [("loss", "val_loss", "Loss"), ("accuracy", "val_accuracy", "Accuracy")],
    ):
        for key, label, color in [(tr, "Train", "#185FA5"), (vl, "Val", "#e74c3c")]:
            mean_v = np.nanmean(mats[key], axis=0)
            std_v = np.nanstd(mats[key], axis=0)
            ax.plot(ep, mean_v, color=color, label=label, linewidth=1.8)
            ax.fill_between(ep, mean_v - std_v, mean_v + std_v, alpha=0.15, color=color)
        ax.set_xlabel("Epoch")
        ax.set_ylabel(ylabel)
        ax.set_title(f"All-fold {ylabel} (mean +/- std)", fontweight="bold")
        ax.legend()
        ax.grid(True, alpha=0.2)
    plt.tight_layout()
    out = os.path.join(results_dir, "training_curves_all_folds.png")
    plt.savefig(out, dpi=150)
    plt.close(fig)
    print(f"  All-fold training curves -> {out}")


def build_cnn_lstm(lr=1e-3, export_compatible=False):
    inp = layers.Input(shape=(WIN_SIZE, N_CHANNELS))
    x = layers.Conv1D(64, 5, padding="same", activation="relu", name="conv1d_1")(inp)
    x = layers.BatchNormalization()(x)
    x = layers.Conv1D(128, 5, padding="same", activation="relu", name="conv1d_2")(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling1D(2)(x)
    x = layers.Dropout(0.3)(x)
    x = layers.Conv1D(128, 3, padding="same", activation="relu", name="conv1d_3")(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling1D(2)(x)
    x = layers.Dropout(0.3)(x)
    if export_compatible:
        x = layers.RNN(layers.LSTMCell(64), name="temporal_lstm")(x)
    else:
        x = layers.LSTM(64, name="temporal_lstm")(x)
    x = layers.Dropout(0.4)(x)
    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dropout(0.3)(x)
    out = layers.Dense(N_GESTURES, activation="softmax")(x)
    model = models.Model(inp, out, name="CNN_LSTM_Gesture")
    model.compile(
        optimizer=tf.keras.optimizers.Adam(lr),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )
    return model


def save_model_summary_text(model, results_dir):
    stream = io.StringIO()
    model.summary(print_fn=lambda x: stream.write(x + "\n"))
    with open(os.path.join(results_dir, "model_summary.txt"), "w", encoding="utf-8") as f:
        f.write(stream.getvalue())


def save_architecture_rationale(results_dir):
    with open(os.path.join(results_dir, "architecture_rationale.txt"), "w", encoding="utf-8") as f:
        f.write(ARCH_RATIONALE.strip() + "\n")


def run_feature_baseline(all_data, results_dir):
    """Leakage-safe handcrafted-feature LOSO baseline."""
    print_banner("FEATURE BASELINE BENCHMARK (HANDCRAFTED FEATURES + LOGISTIC REGRESSION)")

    feature_data = {}
    for sid, (x_sub, y_sub) in all_data.items():
        feature_data[sid] = (extract_handcrafted_features(x_sub), y_sub.copy())

    baseline_rows = []
    all_true = []
    all_pred = []

    subjects = sorted(feature_data.keys())
    for fold_id, test_sid in enumerate(subjects, start=1):
        outer_train_sids = [s for s in subjects if s != test_sid]
        x_train = np.concatenate([feature_data[s][0] for s in outer_train_sids], axis=0)
        y_train = np.concatenate([feature_data[s][1] for s in outer_train_sids], axis=0)
        x_test, y_test = feature_data[test_sid]

        scaler = StandardScaler()
        x_train = scaler.fit_transform(x_train)
        x_test = scaler.transform(x_test)

        clf = LogisticRegression(
            max_iter=500,
            solver="lbfgs",
        )
        clf.fit(x_train, y_train)
        y_pred = clf.predict(x_test)
        acc = float(np.mean(y_pred == y_test))

        baseline_rows.append(
            {
                "fold": fold_id,
                "test_subject": test_sid,
                "accuracy": acc,
            }
        )
        all_true.extend(y_test.tolist())
        all_pred.extend(y_pred.tolist())

        print(f"  Fold {fold_id:02d} | test={test_sid:02d} | acc={acc:.4f}")

    baseline_df = pd.DataFrame(baseline_rows)
    baseline_df.to_csv(os.path.join(results_dir, "baseline_feature_loso_results.csv"), index=False)

    overall_acc = float(np.mean(np.array(all_true) == np.array(all_pred)))
    overall_f1 = float(f1_score(all_true, all_pred, average="weighted"))

    fig, ax = plt.subplots(figsize=(14, 4))
    ax.bar(
        baseline_df["test_subject"],
        baseline_df["accuracy"] * 100,
        color="#7f8c8d",
        edgecolor="black",
        linewidth=0.4,
    )
    ax.axhline(overall_acc * 100, color="black", linestyle="--", linewidth=1.5, label=f"Mean {overall_acc * 100:.1f}%")
    ax.set_xlabel("Subject ID")
    ax.set_ylabel("Accuracy (%)")
    ax.set_title("Feature Baseline -- LOSO Accuracy by Subject", fontweight="bold")
    ax.set_xticks(baseline_df["test_subject"])
    ax.grid(axis="y", alpha=0.3)
    ax.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(results_dir, "baseline_feature_subject_accuracy.png"), dpi=150)
    plt.close(fig)

    summary = {
        "model": "LogisticRegression on handcrafted features",
        "overall_accuracy": overall_acc,
        "weighted_f1": overall_f1,
        "n_subjects": len(subjects),
        "n_features": int(feature_data[subjects[0]][0].shape[1]),
    }
    save_json(os.path.join(results_dir, "baseline_feature_summary.json"), summary)

    print(f"\n  Feature baseline accuracy: {overall_acc * 100:.1f}% | F1: {overall_f1:.4f}")
    return summary


def get_last_conv1d_layer_name(model):
    for layer in reversed(model.layers):
        if isinstance(layer, layers.Conv1D):
            return layer.name
    raise ValueError("No Conv1D layer found for Grad-CAM.")


def compute_gradcam_1d(model, sample, class_idx=None, conv_layer_name=None):
    """Compute 1D Grad-CAM for one sample window."""
    if conv_layer_name is None:
        conv_layer_name = get_last_conv1d_layer_name(model)

    grad_model = tf.keras.models.Model(
        model.inputs,
        [model.get_layer(conv_layer_name).output, model.output],
    )

    x = tf.convert_to_tensor(sample[np.newaxis, ...], dtype=tf.float32)
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(x, training=False)
        if class_idx is None:
            class_idx = int(tf.argmax(predictions[0]))
        target = predictions[:, class_idx]

    grads = tape.gradient(target, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=1)
    conv_outputs = conv_outputs[0]
    pooled_grads = pooled_grads[0]

    cam = tf.reduce_sum(conv_outputs * pooled_grads, axis=-1)
    cam = tf.nn.relu(cam)
    cam = cam.numpy()
    if np.max(cam) > 0:
        cam = cam / np.max(cam)

    if len(cam) != sample.shape[0]:
        src_x = np.linspace(0, 1, len(cam))
        dst_x = np.linspace(0, 1, sample.shape[0])
        cam = np.interp(dst_x, src_x, cam)

    return cam.astype(np.float32), class_idx


def plot_gradcam_examples(model, x_test, y_test, y_pred, y_prob, results_dir, fold_id, test_subject):
    """Create a representative Grad-CAM figure on correctly classified windows."""
    correct_idx = np.where(y_test == y_pred)[0]
    if len(correct_idx) == 0:
        print("  XAI skipped: no correctly classified windows in representative fold.")
        return None

    selected = []
    used_classes = set()
    ranked = correct_idx[np.argsort(np.max(y_prob[correct_idx], axis=1))[::-1]]
    for idx in ranked:
        cls = int(y_test[idx])
        if cls not in used_classes:
            selected.append(int(idx))
            used_classes.add(cls)
        if len(selected) == 6:
            break

    if len(selected) == 0:
        print("  XAI skipped: could not select representative windows.")
        return None

    conv_layer_name = get_last_conv1d_layer_name(model)
    fig, axes = plt.subplots(len(selected), 1, figsize=(14, 2.6 * len(selected)), sharex=True)
    if len(selected) == 1:
        axes = [axes]

    rows = []
    for ax, idx in zip(axes, selected):
        sample = x_test[idx]
        cam, pred_class = compute_gradcam_1d(model, sample, class_idx=int(y_pred[idx]), conv_layer_name=conv_layer_name)
        motion = np.mean(np.abs(sample), axis=1)
        motion = motion / max(np.max(motion), 1e-8)
        t = np.arange(len(motion))

        ax.plot(t, motion, color="black", linewidth=1.5, label="Mean |signal|")
        ax.fill_between(t, 0, cam, color="#e74c3c", alpha=0.35, label="Grad-CAM")
        ax.set_ylim(0, 1.05)
        ax.set_ylabel("Norm.")
        ax.grid(True, alpha=0.2)
        conf = float(np.max(y_prob[idx]))
        ax.set_title(
            f"True={GESTURE_NAMES[int(y_test[idx])]} | Pred={GESTURE_NAMES[pred_class]} | Conf={conf:.3f}",
            fontsize=10,
            fontweight="bold",
        )
        rows.append(
            {
                "index": int(idx),
                "true_class": int(y_test[idx]),
                "pred_class": int(pred_class),
                "confidence": conf,
            }
        )

    axes[0].legend(loc="upper right", fontsize=8)
    axes[-1].set_xlabel("Time Index")
    plt.suptitle(
        f"1D Grad-CAM -- Representative Fold {fold_id:02d} | Test Subject {test_subject:02d}",
        fontsize=13,
        fontweight="bold",
    )
    plt.tight_layout()
    out = os.path.join(results_dir, "xai_gradcam_examples.png")
    plt.savefig(out, dpi=150, bbox_inches="tight")
    plt.close(fig)
    pd.DataFrame(rows).to_csv(os.path.join(results_dir, "xai_gradcam_examples.csv"), index=False)
    print(f"  Grad-CAM figure -> {out}")
    return out


def plot_crossfold_gradcam(cam_accumulator, results_dir):
    """Plot average cross-fold temporal saliency per gesture."""
    fig, axes = plt.subplots(3, 4, figsize=(20, 10))
    axes = axes.flatten()
    for g in range(N_GESTURES):
        ax = axes[g]
        if cam_accumulator[g]:
            mean_cam = np.mean(cam_accumulator[g], axis=0)
            t = np.arange(len(mean_cam))
            ax.fill_between(t, 0, mean_cam, color="#e74c3c", alpha=0.55)
            ax.plot(t, mean_cam, color="#a32d2d", linewidth=1.2)
        ax.set_title(GESTURE_NAMES[g], fontsize=10, fontweight="bold")
        ax.set_ylim(0, 1.05)
        ax.grid(True, alpha=0.2)
        ax.text(
            0.02,
            0.92,
            f"n={len(cam_accumulator[g])}",
            transform=ax.transAxes,
            fontsize=8,
            color="gray",
        )
    axes[11].axis("off")
    plt.suptitle(
        "Cross-subject Grad-CAM saliency per gesture\n"
        "(qualitative temporal saliency - not causal explanation)",
        fontsize=13,
        fontweight="bold",
    )
    plt.tight_layout()
    out = os.path.join(results_dir, "xai_gradcam_crossfold.png")
    plt.savefig(out, dpi=150)
    plt.close(fig)
    print(f"  Cross-fold Grad-CAM -> {out}")


def make_rep_dataset(x_sample, n=200):
    """Representative dataset generator for post-training quantization."""
    n_take = min(len(x_sample), n)
    rng = np.random.default_rng(2027)
    idx = rng.choice(len(x_sample), n_take, replace=False)
    sample = x_sample[idx]

    def gen():
        for w in sample:
            yield [w[np.newaxis, ...].astype(np.float32)]

    return gen


def build_export_compatible_twin(trained_model):
    """Clone weights into a non-CuDNN recurrent graph for TFLite export."""
    export_model = build_cnn_lstm(export_compatible=True)
    export_model(np.zeros((1, WIN_SIZE, N_CHANNELS), dtype=np.float32))
    export_model.set_weights(trained_model.get_weights())
    return export_model


def benchmark_tflite_latency(model_content, reference_window, n_runs=100):
    interpreter = tf.lite.Interpreter(model_content=model_content)
    interpreter.allocate_tensors()
    input_details = interpreter.get_input_details()
    output_details = interpreter.get_output_details()

    sample = reference_window[np.newaxis, ...].astype(np.float32)
    interpreter.set_tensor(input_details[0]["index"], sample)
    interpreter.invoke()
    _ = interpreter.get_tensor(output_details[0]["index"])

    lat_times = []
    for _ in range(n_runs):
        t0 = time.perf_counter()
        interpreter.set_tensor(input_details[0]["index"], sample)
        interpreter.invoke()
        _ = interpreter.get_tensor(output_details[0]["index"])
        lat_times.append((time.perf_counter() - t0) * 1000)
    return float(np.median(lat_times))


def export_quantized_model(model, reference_window, reference_x_train, results_dir):
    """Export dynamic and int8-quantized TFLite variants and summarize them."""
    keras_path = os.path.join(results_dir, "representative_fold_model.keras")
    model.save(keras_path)

    summary = {
        "keras_model_path": keras_path,
        "keras_model_size_kb": round(os.path.getsize(keras_path) / 1024.0, 2),
        "export_status": "pending",
        "export_bridge": "Export-compatible RNN(LSTMCell) twin used to avoid GPU CuDNN-only LSTM ops during TFLite conversion.",
    }

    try:
        export_model = build_export_compatible_twin(model)

        converter = tf.lite.TFLiteConverter.from_keras_model(export_model)
        converter.optimizations = [tf.lite.Optimize.DEFAULT]
        tflite_model_dynamic = converter.convert()

        tflite_path = os.path.join(results_dir, "representative_fold_model_dynamic.tflite")
        with open(tflite_path, "wb") as f:
            f.write(tflite_model_dynamic)

        dynamic_latency = benchmark_tflite_latency(tflite_model_dynamic, reference_window, n_runs=100)

        converter_int8 = tf.lite.TFLiteConverter.from_keras_model(export_model)
        converter_int8.optimizations = [tf.lite.Optimize.DEFAULT]
        converter_int8.representative_dataset = make_rep_dataset(reference_x_train)
        converter_int8.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
        converter_int8.inference_input_type = tf.float32
        converter_int8.inference_output_type = tf.float32
        tflite_model_int8 = converter_int8.convert()

        int8_path = os.path.join(results_dir, "representative_fold_model_int8.tflite")
        with open(int8_path, "wb") as f:
            f.write(tflite_model_int8)

        int8_latency = benchmark_tflite_latency(tflite_model_int8, reference_window, n_runs=100)

        summary.update(
            {
                "export_status": "success",
                "dynamic_tflite_path": tflite_path,
                "dynamic_tflite_size_kb": round(os.path.getsize(tflite_path) / 1024.0, 2),
                "dynamic_tflite_median_latency_ms": round(dynamic_latency, 3),
                "int8_tflite_path": int8_path,
                "int8_tflite_size_kb": round(os.path.getsize(int8_path) / 1024.0, 2),
                "int8_tflite_median_latency_ms": round(int8_latency, 3),
                "int8_quantization_mode": "INT8 internal quantization with float32 input/output",
            }
        )
        print(f"  Quantized TFLite models -> {tflite_path} | {int8_path}")
    except Exception as exc:
        summary.update(
            {
                "export_status": "failed",
                "error_type": type(exc).__name__,
                "error_message": str(exc),
                "traceback": traceback.format_exc(),
                "note": "Training and evaluation remain valid. Only the optional post-training export step failed.",
            }
        )
        print(f"  Quantization export skipped: {type(exc).__name__}: {exc}")

    save_json(os.path.join(results_dir, "quantization_summary.json"), summary)
    return summary


def build_streaming_demo_rows(y_true, y_pred, y_prob, entropy, gate, gesture_to_command):
    rows = []
    for i in range(len(y_true)):
        pred_idx = int(y_pred[i])
        conf = float(np.max(y_prob[i]))
        accepted = conf >= gate
        rows.append(
            {
                "sample_index": i,
                "true_gesture": GESTURE_NAMES[int(y_true[i])],
                "pred_gesture": GESTURE_NAMES[pred_idx],
                "max_confidence": round(conf, 4),
                "entropy": round(float(entropy[i]), 4),
                "accepted": bool(accepted),
                "assigned_command": gesture_to_command[pred_idx] if accepted else "MANUAL_REVIEW",
            }
        )
    return rows


def compute_ece_and_plot(y_true, y_pred, y_prob, results_dir, n_bins=15):
    """Top-label reliability diagram with ECE and MCE."""
    max_conf = np.max(y_prob, axis=1)
    correct = (y_true == y_pred).astype(float)
    bins = np.linspace(0, 1, n_bins + 1)
    bin_acc, bin_conf, bin_count = [], [], []

    for lo, hi in zip(bins[:-1], bins[1:]):
        mask = (max_conf >= lo) & (max_conf < hi)
        if hi == 1.0:
            mask = (max_conf >= lo) & (max_conf <= hi)
        if mask.sum() == 0:
            continue
        bin_acc.append(float(correct[mask].mean()))
        bin_conf.append(float(max_conf[mask].mean()))
        bin_count.append(int(mask.sum()))

    bin_acc = np.array(bin_acc, dtype=np.float32)
    bin_conf = np.array(bin_conf, dtype=np.float32)
    bin_count = np.array(bin_count, dtype=np.int64)

    ece = float(np.sum(bin_count * np.abs(bin_acc - bin_conf)) / len(y_true))
    mce = float(np.max(np.abs(bin_acc - bin_conf))) if len(bin_acc) else 0.0

    fig, ax = plt.subplots(figsize=(7, 6))
    ax.bar(
        bin_conf,
        bin_acc,
        width=0.05,
        alpha=0.7,
        color="#185FA5",
        edgecolor="black",
        linewidth=0.5,
        label="Model",
    )
    ax.plot([0, 1], [0, 1], "k--", linewidth=1.5, label="Perfect calibration")
    ax.set_xlabel("Mean predicted confidence")
    ax.set_ylabel("Fraction correct")
    ax.set_title(f"Reliability diagram | ECE={ece:.4f} | MCE={mce:.4f}", fontweight="bold")
    ax.legend()
    ax.grid(True, alpha=0.2)
    plt.tight_layout()
    plt.savefig(os.path.join(results_dir, "calibration_reliability.png"), dpi=150)
    plt.close(fig)

    save_json(
        os.path.join(results_dir, "calibration_summary.json"),
        {"ece": round(ece, 6), "mce": round(mce, 6), "n_bins": n_bins},
    )
    print(f"  Calibration -> ECE={ece:.4f} | MCE={mce:.4f}")
    return ece, mce


def save_final_project_summary(results_dir, summary_payload):
    md = f"""# HGAG Final Project Summary

## Final Performance

- LOSO accuracy: **{summary_payload['overall_accuracy_pct']:.2f}%**
- Weighted F1: **{summary_payload['weighted_f1']:.4f}**
- Mean ROC-AUC: **{summary_payload['mean_auc']:.4f}**
- Preprocessing mode: **{summary_payload['preprocess_mode']}**

## Evaluation Integrity

- Subject-pure LOSO evaluation across 43 subjects
- Validation subjects selected from training-side subjects only
- StandardScaler fit on training data only
- Augmentation applied to training windows only
- Test data never used for early stopping or learning-rate scheduling

## Deployment Readiness

- Recommended confidence gate: **{CONFIDENCE_GATE:.2f}**
- Gated accuracy at {CONFIDENCE_GATE:.2f}: **{summary_payload['gated_accuracy_at_0_70_pct']:.2f}%**
- Mean latency: **{summary_payload['latency_mean_ms']:.2f} ms**
- Safe-assignment risk reduction: **{summary_payload['risk_reduction_pct']:.2f}%**
- Calibration ECE: **{summary_payload['calibration_ece']:.4f}**

## Academic Additions

- Runtime-computed preprocessing audit justification
- Handcrafted-feature LOSO benchmark
- 1D Grad-CAM interpretability on a representative fold
- Cross-fold Grad-CAM saliency aggregation
- Dynamic and representative-dataset int8 TFLite quantization export
- Executive dashboard and deployment diagnostics

## Ethical Considerations

- **Asymmetric error cost**: A false accept on EMERGENCY STOP or Start Machine
  carries physical risk disproportionate to general misclassification.
  The Hungarian assignment engine mitigates this by mapping low-error-rate
  gestures to high-criticality commands, but does not eliminate the risk.

- **Subject variability and fairness**: Model performance varies across subjects.
  Deployment populations with motion profiles outside the training distribution
  may see degraded accuracy. Revalidation on deployment cohorts is recommended.

- **Confidence gating as a human override mechanism**: The {CONFIDENCE_GATE:.2f} gate is a
  minimum operating threshold, not a guarantee of safety. Higher-criticality
  commands should use stricter gates and fallback review logic.

- **No auto-execution of high-criticality commands**: Commands with criticality
  >= 8 should require a secondary confirmation step before actuation,
  regardless of model confidence.
"""
    with open(os.path.join(results_dir, "FINAL_PROJECT_SUMMARY.md"), "w", encoding="utf-8") as f:
        f.write(md)


def save_rubric_coverage(results_dir, payload):
    save_json(os.path.join(results_dir, "rubric_coverage_summary.json"), payload)


def save_project_metadata(runtime_audit_summary_file=None):
    metadata = {
        "project_stage": "final_submission_notebook",
        "preprocess_mode": PREPROCESS_MODE,
        "sampling_rate_hz": SR,
        "window_offset_samples": WINDOW_OFFSET,
        "window_size_samples": WIN_SIZE,
        "step_size_samples": STEP_SIZE,
        "val_subjects_per_fold": VAL_SUBJECTS_PER_FOLD,
        "epochs": EPOCHS,
        "batch_size": BATCH,
        "patience": PATIENCE,
        "gesture_names": GESTURE_NAMES,
        "channel_names": CHANNEL_NAMES,
        "augmentation_config": AUG_CONFIG,
        "segmentation_design": {
            "window_offset_samples": WINDOW_OFFSET,
            "window_offset_ms": int(round(WINDOW_OFFSET / SR * 1000.0)),
            "window_size_samples": WIN_SIZE,
            "window_size_ms": int(round(WIN_SIZE / SR * 1000.0)),
            "step_size_samples": STEP_SIZE,
            "step_size_ms": int(round(STEP_SIZE / SR * 1000.0)),
            "segmentation_rule": "Trim first 250 ms, keep the remaining 200 samples as one window per trial.",
        },
        "augmentation_design": {
            "placement_in_pipeline": [
                "load released signal",
                "apply chosen segmentation/windowing",
                "split train/validation/test",
                "fit StandardScaler on training windows only",
                "transform validation/test with train-fitted scaler",
                "apply augmentation to training windows only",
            ],
            "candidate_augmentations": [
                "Gaussian noise jitter",
                "Amplitude scaling",
                "Window time warping",
            ],
            "notes": [
                "Validation and test data remain unaugmented.",
                "Augmentation choice is validated empirically, not assumed from preprocessing audit evidence.",
            ],
        },
        "runtime_audit_summary_file": runtime_audit_summary_file,
        "rubric_focus": {
            "problem_understanding": True,
            "model_design": True,
            "preprocessing_and_feature_engineering": True,
            "training_and_evaluation": True,
            "interpretability": True,
            "deployment_readiness": True,
            "xai_bonus": RUN_XAI,
            "model_optimization_bonus": RUN_MODEL_OPTIMIZATION,
            "creative_dashboard_bonus": True,
        },
    }
    save_json(os.path.join(RESULTS_DIR, "project_run_metadata.json"), metadata)


os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(RESULTS_DIR, exist_ok=True)
tf.keras.utils.set_random_seed(42)
save_project_metadata()

print_banner("HGAG FINAL SUBMISSION CONFIGURATION")
print(f"  Sampling rate:        {SR} Hz")
print(f"  Preprocess mode:      {PREPROCESS_MODE}")
print(f"  Window offset:        {WINDOW_OFFSET} samples ({WINDOW_OFFSET / SR * 1000:.0f} ms)")
print(f"  Window size:          {WIN_SIZE} samples ({WIN_SIZE / SR:.3f} s)")
print(f"  Step size:            {STEP_SIZE} samples (1 window/trial)")
print(f"  Validation subjects:  {VAL_SUBJECTS_PER_FOLD} per fold")
print(f"  Epochs:               {EPOCHS}")
print(f"  Batch size:           {BATCH}")
print(f"  Patience:             {PATIENCE}")
print(f"  Baseline benchmark:   {RUN_BASELINE_BENCHMARK}")
print(f"  XAI enabled:          {RUN_XAI}")
print(f"  Quantization enabled: {RUN_MODEL_OPTIMIZATION}")
print(f"  Streaming demo:       {RUN_STREAMING_DEMO}")


# -----------------------------------------------------------------------------
# 1. Problem Understanding & Dataset Analysis
# -----------------------------------------------------------------------------
print_banner("PROBLEM UNDERSTANDING & DATASET ANALYSIS")
print("Loading all subjects with audit-backed preprocessing...")

all_data = {}
for sid in range(1, N_SUBJECTS + 1):
    x_sub, y_sub = load_subject(sid)
    if x_sub is not None:
        all_data[sid] = (x_sub, y_sub)
        print(
            f"  Subject {sid:02d}: {len(x_sub):5d} windows | "
            f"classes: {sorted(np.unique(y_sub).tolist())}"
        )
    else:
        print(f"  Subject {sid:02d}: SKIPPED")

subjects = sorted(all_data.keys())
if len(subjects) == 0:
    raise RuntimeError(f"No subjects loaded. Check RAW_PATH: {RAW_PATH}")

plot_dataset_overview(all_data, RESULTS_DIR)
plot_raw_window_visuals(all_data, RESULTS_DIR)
plot_released_signal_example(RESULTS_DIR)
runtime_audit_summary = compute_runtime_audit_summary(RESULTS_DIR)
save_project_metadata(runtime_audit_summary_file="runtime_audit_summary.json")
class_balance_summary = report_class_balance(all_data, RESULTS_DIR)

total_windows = int(sum(len(v[0]) for v in all_data.values()))
print(f"\nLoaded {len(subjects)} subjects and {total_windows:,} segmented windows.")
print("Runtime audit notes:")
print(
    f"  - Trials are fixed at {runtime_audit_summary['trial_length_samples']} samples "
    f"({runtime_audit_summary['trial_length_ms']} ms at {SR} Hz)."
)
print(
    "  - Median quiet lead-in ratios by channel were computed in this run and "
    "saved to runtime_audit_summary.json."
)
print("  - Released signal remains the default final preprocessing choice.")


# -----------------------------------------------------------------------------
# 2. Data Preprocessing & Feature Engineering
# -----------------------------------------------------------------------------
baseline_summary = None
if RUN_BASELINE_BENCHMARK:
    baseline_summary = run_feature_baseline(all_data, RESULTS_DIR)


# -----------------------------------------------------------------------------
# 3. Fold preparation
# -----------------------------------------------------------------------------
print_banner("FOLD GENERATION WITH LEAKAGE-SAFE PREPROCESSING")

fold_manifest_rows = []
aug_preview_saved = False

for fold_id, test_sid in enumerate(subjects, start=1):
    outer_train_sids = [s for s in subjects if s != test_sid]
    train_sids, val_sids = choose_validation_subjects(outer_train_sids, fold_id)

    x_train_raw = np.concatenate([all_data[s][0] for s in train_sids])
    y_train_raw = np.concatenate([all_data[s][1] for s in train_sids])
    x_val_raw = np.concatenate([all_data[s][0] for s in val_sids])
    y_val_raw = np.concatenate([all_data[s][1] for s in val_sids])
    x_test_raw, y_test = all_data[test_sid]

    scaler = StandardScaler()
    n_train, t, c = x_train_raw.shape
    n_val = x_val_raw.shape[0]
    n_test = x_test_raw.shape[0]

    x_train = scaler.fit_transform(x_train_raw.reshape(-1, c)).reshape(n_train, t, c).astype(np.float32)
    x_val = scaler.transform(x_val_raw.reshape(-1, c)).reshape(n_val, t, c).astype(np.float32)
    x_test = scaler.transform(x_test_raw.reshape(-1, c)).reshape(n_test, t, c).astype(np.float32)

    if not aug_preview_saved:
        plot_augmentation_preview(x_train[0], RESULTS_DIR)
        aug_preview_saved = True

    x_calib = x_train[: min(len(x_train), 256)].copy()
    fold_rng = np.random.default_rng(42 + fold_id)
    x_train_aug, y_train_aug = augment(x_train, y_train_raw, fold_rng)

    np.savez_compressed(
        os.path.join(OUTPUT_DIR, f"fold_{fold_id:02d}.npz"),
        x_train=x_train_aug,
        y_train=y_train_aug,
        x_val=x_val,
        y_val=y_val_raw,
        x_test=x_test,
        y_test=y_test,
        x_calib=x_calib,
        fold=np.array([fold_id], dtype=np.int32),
        test_subject=np.array([test_sid], dtype=np.int32),
        train_subjects=np.array(train_sids, dtype=np.int32),
        val_subjects=np.array(val_sids, dtype=np.int32),
        preprocess_mode=np.array([PREPROCESS_MODE]),
    )

    fold_manifest_rows.append(
        {
            "fold": fold_id,
            "test_subject": test_sid,
            "n_train_subjects": len(train_sids),
            "n_val_subjects": len(val_sids),
            "train_windows_augmented": len(x_train_aug),
            "val_windows": len(x_val),
            "test_windows": len(x_test),
            "preprocess_mode": PREPROCESS_MODE,
        }
    )

    print(
        f"Fold {fold_id:02d} | test={test_sid:02d} | "
        f"train_subj={len(train_sids):02d} | val_subj={len(val_sids):02d} | "
        f"train={len(x_train_aug):6d} | val={len(x_val):4d} | test={len(x_test):4d} OK"
    )

    del x_train, y_train_raw, x_val, y_val_raw, x_test, y_test
    del x_train_raw, x_val_raw, x_test_raw, x_train_aug, y_train_aug
    gc.collect()

pd.DataFrame(fold_manifest_rows).to_csv(os.path.join(RESULTS_DIR, "fold_manifest.csv"), index=False)

N_FOLDS = len(subjects)
print(f"\nPrepared {N_FOLDS} LOSO folds.")


# -----------------------------------------------------------------------------
# 4. Model Design & Justification / Training
# -----------------------------------------------------------------------------
print_banner("MODEL DESIGN & TRAINING")

model = build_cnn_lstm()
save_model_summary_text(model, RESULTS_DIR)
save_architecture_rationale(RESULTS_DIR)
model.summary()
print_banner("MODEL ARCHITECTURE RATIONALE")
print(ARCH_RATIONALE)

RESULTS_JSON = os.path.join(RESULTS_DIR, "loso_results.json")
PREDS_NPZ = os.path.join(RESULTS_DIR, "predictions.npz")
FOLD_SUMMARY_CSV = os.path.join(RESULTS_DIR, "loso_fold_summary.csv")
ALL_HISTORIES_JSON = os.path.join(RESULTS_DIR, "all_fold_histories.json")
XAI_CAM_JSON = os.path.join(RESULTS_DIR, "xai_cam_accumulator.json")

completed_folds = set()
all_results = []
all_y_true, all_y_pred, all_y_prob = [], [], []
representative_assets = {}
all_histories = []
cam_accumulator = {g: [] for g in range(N_GESTURES)}

if os.path.exists(RESULTS_JSON):
    with open(RESULTS_JSON, encoding="utf-8") as f:
        all_results = json.load(f)
    completed_folds = {r["fold"] for r in all_results}
    if not os.path.exists(PREDS_NPZ):
        raise RuntimeError(
            "Resume state is incomplete: loso_results.json exists but predictions.npz is missing."
        )
    prev = np.load(PREDS_NPZ)
    all_y_true = prev["y_true"].tolist()
    all_y_pred = prev["y_pred"].tolist()
    all_y_prob = prev["y_prob"].tolist()
    del prev
    if not (len(all_y_true) == len(all_y_pred) == len(all_y_prob)):
        raise RuntimeError(
            f"Resume mismatch: y_true={len(all_y_true)}, y_pred={len(all_y_pred)}, y_prob={len(all_y_prob)}"
        )
    expected_n = sum(len(r["y_true"]) for r in all_results)
    if not (len(all_y_true) == len(all_y_pred) == len(all_y_prob) == expected_n):
        raise RuntimeError(
            "Resume mismatch between loso_results.json and predictions.npz: "
            f"expected {expected_n}, got {len(all_y_true)} / {len(all_y_pred)} / {len(all_y_prob)}"
        )
    if os.path.exists(ALL_HISTORIES_JSON):
        with open(ALL_HISTORIES_JSON, encoding="utf-8") as f:
            all_histories = json.load(f)
    if RUN_XAI and os.path.exists(XAI_CAM_JSON):
        with open(XAI_CAM_JSON, encoding="utf-8") as f:
            raw_cam = json.load(f)
        cam_accumulator = {int(k): v for k, v in raw_cam.items()}
    print(f"Resuming: {len(completed_folds)} folds already complete.")
else:
    if os.path.exists(PREDS_NPZ):
        raise RuntimeError(
            "Resume state is incomplete: predictions.npz exists but loso_results.json is missing."
        )
    print("Starting fresh -- no previous results found.")

t_total = time.perf_counter()

for fold_id in range(1, N_FOLDS + 1):
    if fold_id in completed_folds:
        prev = next(r for r in all_results if r["fold"] == fold_id)
        print(f"Fold {fold_id:02d} | SKIPPED (already done) | acc={prev['accuracy']:.4f}")
        continue

    data = np.load(os.path.join(OUTPUT_DIR, f"fold_{fold_id:02d}.npz"))
    x_train, y_train = data["x_train"], data["y_train"]
    x_val, y_val = data["x_val"], data["y_val"]
    x_test, y_test = data["x_test"], data["y_test"]
    x_calib = data["x_calib"] if "x_calib" in data.files else x_train[: min(len(x_train), 256)]
    test_sid = int(data["test_subject"][0])
    val_sids = data["val_subjects"].tolist()

    print(
        f"Fold {fold_id:02d} | train: {len(x_train)} | "
        f"val: {len(x_val)} | test: {len(x_test)} | val_subjects={val_sids}"
    )

    t0 = time.perf_counter()
    model = build_cnn_lstm()
    history = model.fit(
        x_train,
        y_train,
        validation_data=(x_val, y_val),
        epochs=EPOCHS,
        batch_size=BATCH,
        callbacks=[
            cb.EarlyStopping(
                monitor="val_loss",
                patience=PATIENCE,
                restore_best_weights=True,
                verbose=0,
            ),
            cb.ReduceLROnPlateau(
                monitor="val_loss",
                factor=0.5,
                patience=3,
                verbose=0,
            ),
        ],
        verbose=2,
    )

    plot_training_curves(history, fold_id, RESULTS_DIR)
    all_histories.append(
        {
            "fold": fold_id,
            "loss": history.history["loss"],
            "val_loss": history.history["val_loss"],
            "accuracy": history.history["accuracy"],
            "val_accuracy": history.history["val_accuracy"],
        }
    )

    y_prob = model.predict(x_test, batch_size=256, verbose=0)
    y_pred = np.argmax(y_prob, axis=1)
    acc = float(np.mean(y_pred == y_test))
    elapsed = time.perf_counter() - t0
    best_val_acc = float(np.max(history.history["val_accuracy"]))

    single_window = x_test[[0]]
    single_window_tf = tf.constant(single_window, dtype=tf.float32)
    _ = model(single_window_tf, training=False)
    lat_times = []
    for _ in range(50):
        t_inf = time.perf_counter()
        _ = model(single_window_tf, training=False)
        lat_times.append((time.perf_counter() - t_inf) * 1000)
    median_lat_ms = float(np.median(lat_times))

    needs_rep_xai = RUN_XAI and not os.path.exists(os.path.join(RESULTS_DIR, "xai_gradcam_examples.png"))
    needs_rep_quant = RUN_MODEL_OPTIMIZATION and not os.path.exists(
        os.path.join(RESULTS_DIR, "quantization_summary.json")
    )
    if needs_rep_xai or needs_rep_quant:
        representative_assets["fold_id"] = fold_id
        representative_assets["test_subject"] = test_sid
        representative_assets["x_test"] = x_test.copy()
        representative_assets["y_test"] = y_test.copy()
        representative_assets["y_pred"] = y_pred.copy()
        representative_assets["y_prob"] = y_prob.copy()
        if needs_rep_xai:
            plot_gradcam_examples(model, x_test, y_test, y_pred, y_prob, RESULTS_DIR, fold_id, test_sid)
        if needs_rep_quant:
            export_quantized_model(model, x_test[0], x_calib, RESULTS_DIR)

    if RUN_XAI:
        conv_name = get_last_conv1d_layer_name(model)
        correct_mask = y_pred == y_test
        for g in range(N_GESTURES):
            g_idx = np.where(correct_mask & (y_test == g))[0]
            if len(g_idx) == 0:
                continue
            best_local = int(g_idx[np.argmax(np.max(y_prob[g_idx], axis=1))])
            cam, _ = compute_gradcam_1d(
                model,
                x_test[best_local],
                class_idx=g,
                conv_layer_name=conv_name,
            )
            cam_accumulator[g].append(cam)
        save_json(
            XAI_CAM_JSON,
            {str(k): [np.asarray(cam).tolist() for cam in v] for k, v in cam_accumulator.items()},
        )

    all_y_true.extend(y_test.tolist())
    all_y_pred.extend(y_pred.tolist())
    all_y_prob.extend(y_prob.tolist())

    epochs_trained = len(history.history["loss"])
    print(
        f"  -> acc={acc:.4f} | best_val_acc={best_val_acc:.4f} | "
        f"epochs={epochs_trained} | lat={median_lat_ms:.1f}ms | {elapsed:.0f}s OK"
    )

    all_results.append(
        {
            "fold": fold_id,
            "test_subject": test_sid,
            "val_subjects": val_sids,
            "accuracy": acc,
            "best_val_accuracy": best_val_acc,
            "epochs_trained": epochs_trained,
            "median_latency_ms": median_lat_ms,
            "preprocess_mode": PREPROCESS_MODE,
            "y_true": y_test.tolist(),
            "y_pred": y_pred.tolist(),
        }
    )

    with open(RESULTS_JSON, "w", encoding="utf-8") as f:
        json.dump(all_results, f)
    save_json(ALL_HISTORIES_JSON, all_histories)
    np.savez_compressed(
        PREDS_NPZ,
        y_true=np.array(all_y_true),
        y_pred=np.array(all_y_pred),
        y_prob=np.array(all_y_prob),
    )

    fold_summary = [
        {
            "fold": r["fold"],
            "test_subject": r["test_subject"],
            "accuracy": r["accuracy"],
            "best_val_accuracy": r["best_val_accuracy"],
            "epochs_trained": r["epochs_trained"],
            "median_latency_ms": r["median_latency_ms"],
            "preprocess_mode": r.get("preprocess_mode", PREPROCESS_MODE),
        }
        for r in all_results
    ]
    pd.DataFrame(fold_summary).sort_values("fold").to_csv(FOLD_SUMMARY_CSV, index=False)

    del model, x_train, y_train, x_val, y_val, x_test, y_test, x_calib, data, y_prob
    tf.keras.backend.clear_session()
    gc.collect()

total_time = time.perf_counter() - t_total
overall_acc = float(np.mean(np.array(all_y_true) == np.array(all_y_pred)))
save_json(os.path.join(RESULTS_DIR, "all_fold_histories.json"), all_histories)
plot_all_fold_curves(all_histories, RESULTS_DIR)
if RUN_XAI:
    plot_crossfold_gradcam(cam_accumulator, RESULTS_DIR)
print(f"\nLOSO complete. Accuracy: {overall_acc * 100:.1f}% | Time: {total_time / 60:.1f} min")


# -----------------------------------------------------------------------------
# 5. Model Training & Evaluation
# -----------------------------------------------------------------------------
print_banner("MODEL TRAINING & EVALUATION")

preds = np.load(PREDS_NPZ)
y_true = preds["y_true"]
y_pred = preds["y_pred"]
y_prob = preds["y_prob"]

overall_acc = accuracy_score(y_true, y_pred)
overall_f1 = f1_score(y_true, y_pred, average="weighted")
print(f"Overall Accuracy: {overall_acc * 100:.1f}% | Weighted F1: {overall_f1:.4f}\n")

cm = confusion_matrix(y_true, y_pred)
cm_pct = cm.astype(float) / cm.sum(axis=1, keepdims=True) * 100
np.save(os.path.join(RESULTS_DIR, "confusion_matrix.npy"), cm)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(22, 9))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    ax=ax1,
    xticklabels=GESTURE_NAMES,
    yticklabels=GESTURE_NAMES,
    linewidths=0.5,
)
ax1.set_title("Confusion Matrix (Counts)")
ax1.set_xlabel("Predicted")
ax1.set_ylabel("True")
ax1.tick_params(axis="x", rotation=45)

sns.heatmap(
    cm_pct,
    annot=True,
    fmt=".1f",
    cmap="Blues",
    ax=ax2,
    xticklabels=GESTURE_NAMES,
    yticklabels=GESTURE_NAMES,
    linewidths=0.5,
    vmin=0,
    vmax=100,
)
ax2.set_title("Confusion Matrix (%)")
ax2.set_xlabel("Predicted")
ax2.set_ylabel("True")
ax2.tick_params(axis="x", rotation=45)

plt.suptitle(
    f"LOSO -- Accuracy: {overall_acc * 100:.1f}% | F1: {overall_f1:.4f}",
    fontsize=14,
    fontweight="bold",
)
plt.tight_layout()
plt.savefig(os.path.join(RESULTS_DIR, "confusion_matrix.png"), dpi=150)
plt.show()

with open(RESULTS_JSON, encoding="utf-8") as f:
    all_results = json.load(f)

subjects_list = [r["test_subject"] for r in all_results]
accs = [r["accuracy"] for r in all_results]
fig, ax = plt.subplots(figsize=(16, 5))
colors = ["#2ecc71" if a >= overall_acc else "#e74c3c" for a in accs]
ax.bar(subjects_list, [a * 100 for a in accs], color=colors, edgecolor="black", linewidth=0.5)
ax.axhline(
    y=overall_acc * 100,
    color="black",
    linestyle="--",
    linewidth=1.5,
    label=f"Mean: {overall_acc * 100:.1f}%",
)
ax.set_xlabel("Subject ID")
ax.set_ylabel("Accuracy (%)")
ax.set_ylim(0, 105)
ax.set_title("Per-Subject Accuracy (LOSO)", fontsize=13, fontweight="bold")
ax.legend()
ax.set_xticks(subjects_list)
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(RESULTS_DIR, "per_subject_accuracy.png"), dpi=150)
plt.show()

print("\n" + classification_report(y_true, y_pred, target_names=GESTURE_NAMES, digits=3))

cm_off = cm.copy()
np.fill_diagonal(cm_off, 0)
pairs = [
    (cm_off[i, j], GESTURE_NAMES[i], GESTURE_NAMES[j])
    for i in range(N_GESTURES)
    for j in range(N_GESTURES)
    if i != j and cm_off[i, j] > 0
]
pairs.sort(reverse=True)
print("\nTop 5 Confused Pairs:")
for cnt, g_true, g_pred in pairs[:5]:
    print(f"  {g_true:20s} -> {g_pred:20s} | {cnt} times")

pd.DataFrame(
    [{"count": cnt, "true_gesture": g_true, "pred_gesture": g_pred} for cnt, g_true, g_pred in pairs[:10]]
).to_csv(os.path.join(RESULTS_DIR, "top_confused_pairs.csv"), index=False)

y_bin = label_binarize(y_true, classes=range(N_GESTURES))
auc_scores = [roc_auc_score(y_bin[:, i], y_prob[:, i]) for i in range(N_GESTURES)]
mean_auc = float(np.mean(auc_scores))

fig, ax = plt.subplots(figsize=(10, 6))
colors_auc = ["#e74c3c" if a < 0.95 else "#2ecc71" for a in auc_scores]
bars = ax.barh(GESTURE_NAMES, auc_scores, color=colors_auc, edgecolor="black", linewidth=0.5)
ax.axvline(x=0.95, color="black", linestyle="--", linewidth=2, label="0.95 deployment threshold")
ax.axvline(x=mean_auc, color="#3498db", linestyle="-", linewidth=1.5, label=f"Mean AUC {mean_auc:.3f}")
for bar, a in zip(bars, auc_scores):
    ax.text(bar.get_width() + 0.002, bar.get_y() + bar.get_height() / 2, f"{a:.3f}", va="center", fontsize=9)
ax.set_title("Per-Gesture ROC-AUC (One-vs-Rest)", fontweight="bold")
ax.set_xlim(0.85, 1.01)
ax.legend()
ax.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(RESULTS_DIR, "roc_auc_per_gesture.png"), dpi=150)
plt.show()

pd.DataFrame({"gesture": GESTURE_NAMES, "roc_auc": auc_scores}).sort_values("roc_auc").to_csv(
    os.path.join(RESULTS_DIR, "roc_auc_per_gesture.csv"), index=False
)

if baseline_summary is not None:
    benchmark_df = pd.DataFrame(
        [
            {
                "model": "Feature Baseline (LogReg)",
                "accuracy_pct": baseline_summary["overall_accuracy"] * 100,
                "weighted_f1": baseline_summary["weighted_f1"],
            },
            {
                "model": "CNN-LSTM",
                "accuracy_pct": overall_acc * 100,
                "weighted_f1": overall_f1,
            },
        ]
    )
    benchmark_df.to_csv(os.path.join(RESULTS_DIR, "benchmark_comparison.csv"), index=False)
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.bar(benchmark_df["model"], benchmark_df["accuracy_pct"], color=["#7f8c8d", "#2ecc71"], edgecolor="black")
    for i, v in enumerate(benchmark_df["accuracy_pct"]):
        ax.text(i, v + 0.5, f"{v:.1f}%", ha="center", fontsize=10, fontweight="bold")
    ax.set_ylabel("Accuracy (%)")
    ax.set_ylim(0, 105)
    ax.set_title("Leakage-Safe Benchmark Comparison", fontweight="bold")
    ax.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.savefig(os.path.join(RESULTS_DIR, "benchmark_comparison.png"), dpi=150)
    plt.close(fig)


# -----------------------------------------------------------------------------
# 6. Deployment Readiness / Application Insight
# -----------------------------------------------------------------------------
print_banner("DEPLOYMENT READINESS / APPLICATION INSIGHT")

max_conf = np.max(y_prob, axis=1)
print("\nConfidence Gating Table (deployable thresholds):")
print(f"  {'Gate':>6} | {'Accepted':>10} | {'Accept%':>8} | {'Gated Acc':>10} | Note")
print(f"  {'-' * 60}")
gating_rows = []
for thresh in [0.50, 0.60, CONFIDENCE_GATE, 0.80, 0.90]:
    mask = max_conf >= thresh
    g_acc = np.mean(y_true[mask] == y_pred[mask]) * 100 if mask.any() else 0
    tag = "  <- RECOMMENDED" if np.isclose(thresh, CONFIDENCE_GATE) else ""
    print(f"  {thresh:.2f}  | {mask.sum():>10,} | {mask.mean() * 100:>7.1f}% | {g_acc:>9.1f}% | {tag}")
    gating_rows.append(
        {
            "threshold": thresh,
            "accepted": int(mask.sum()),
            "accept_pct": round(mask.mean() * 100, 1),
            "gated_acc": round(g_acc, 2),
            "recommended": bool(np.isclose(thresh, CONFIDENCE_GATE)),
        }
    )
gating_df = pd.DataFrame(gating_rows)
gating_df.to_csv(os.path.join(RESULTS_DIR, "gating_table.csv"), index=False)

entropy = -np.sum(y_prob * np.log(y_prob + 1e-9), axis=1)
correct = y_true == y_pred

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ax1.hist(max_conf[correct], bins=60, alpha=0.7, color="#2ecc71", label=f"Correct ({correct.sum():,})", density=True)
ax1.hist(max_conf[~correct], bins=60, alpha=0.7, color="#e74c3c", label=f"Wrong ({(~correct).sum():,})", density=True)
ax1.axvline(CONFIDENCE_GATE, color="black", linestyle="--", linewidth=2, label=f"Gate @ {CONFIDENCE_GATE:.2f}")
ax1.set_xlabel("Max Softmax Confidence")
ax1.set_ylabel("Density")
ax1.set_title("Confidence Distribution", fontweight="bold")
ax1.legend()

ax2.hist(entropy[correct], bins=60, alpha=0.7, color="#2ecc71", label="Correct", density=True)
ax2.hist(entropy[~correct], bins=60, alpha=0.7, color="#e74c3c", label="Wrong", density=True)
ax2.set_xlabel("Prediction Entropy (nats)")
ax2.set_ylabel("Density")
ax2.set_title("Prediction Entropy Distribution", fontweight="bold")
ax2.legend()

plt.suptitle("Model Calibration Analysis -- Deployment Safety", fontweight="bold")
plt.tight_layout()
plt.savefig(os.path.join(RESULTS_DIR, "confidence_entropy.png"), dpi=150)
plt.show()

ece, mce = compute_ece_and_plot(y_true, y_pred, y_prob, RESULTS_DIR, n_bins=15)

adv_mask = (max_conf > 0.85) & (entropy > np.percentile(entropy, 75))
adv_count = int(adv_mask.sum())
adv_pct = float(adv_mask.mean() * 100)
print(f"\nHigh-confidence / high-entropy risk candidates: {adv_count:,} ({adv_pct:.1f}%)")

latencies = [r.get("median_latency_ms", None) for r in all_results]
latencies_valid = [(r["test_subject"], l) for r, l in zip(all_results, latencies) if l is not None]

lat_budget_pass = None
lat_mean = None
lat_max = None
if latencies_valid:
    lat_subjs, lat_vals = zip(*latencies_valid)
    lat_mean = float(np.mean(lat_vals))
    lat_max = float(np.max(lat_vals))
    fig, ax = plt.subplots(figsize=(14, 4))
    ax.bar(lat_subjs, lat_vals, color="#3498db", edgecolor="black", linewidth=0.4)
    ax.axhline(100, color="#e74c3c", linestyle="--", linewidth=2, label="100 ms real-time budget")
    ax.axhline(lat_mean, color="black", linestyle="-", linewidth=1.5, label=f"Mean {lat_mean:.1f} ms")
    ax.set_xlabel("Subject (fold)")
    ax.set_ylabel("Median Inference Latency (ms)")
    ax.set_title("Inference Latency per Fold vs 100 ms Real-Time Budget", fontweight="bold")
    ax.set_ylim(0, max(max(lat_vals) * 1.4, 120))
    ax.legend()
    ax.grid(axis="y", alpha=0.3)
    ax.set_xticks(lat_subjs)
    plt.tight_layout()
    plt.savefig(os.path.join(RESULTS_DIR, "inference_latency.png"), dpi=150)
    plt.show()
    lat_budget_pass = int(sum(l < 100 for l in lat_vals))
    print(f"Latency budget: {lat_budget_pass}/{len(lat_vals)} folds under 100 ms")
    print(f"Mean: {lat_mean:.1f} ms | Max: {lat_max:.1f} ms")

cm_rate = cm.astype(float) / cm.sum(axis=1, keepdims=True)
error_rates = 1.0 - np.diag(cm_rate)
criticality = np.array([c["criticality"] for c in COMMANDS], dtype=float)
cost_matrix = np.outer(criticality, error_rates)

row_ind, col_ind = linear_sum_assignment(cost_matrix)
optimal_cost = float(cost_matrix[row_ind, col_ind].sum())
naive_cost = float(cost_matrix[np.arange(N_GESTURES), np.arange(N_GESTURES)].sum())
_, worst_col = linear_sum_assignment(-cost_matrix)
worst_cost = float(cost_matrix[np.arange(N_GESTURES), worst_col].sum())
improvement = float((1 - optimal_cost / naive_cost) * 100) if naive_cost > 0 else 0.0

print("\n" + "=" * 70)
print("  SAFE GESTURE-COMMAND ASSIGNMENT ENGINE")
print("=" * 70)
print(f"\n  Naive cost: {naive_cost:.4f} | Optimal cost: {optimal_cost:.4f} | Risk reduction: {improvement:.1f}%\n")
print(f"{'-' * 70}")
print(f"  {'Command':<22} {'Crit':>4} {'Gesture':<22} {'Error':>7}")
print(f"{'-' * 70}")

assignment_rows = []
gesture_to_command = {}
for cmd_i, gest_i in zip(row_ind, col_ind):
    tag = "REVIEW" if error_rates[gest_i] > 0.10 else "OK"
    gesture_to_command[gest_i] = COMMANDS[cmd_i]["name"]
    print(
        f"  {COMMANDS[cmd_i]['name']:<22} {COMMANDS[cmd_i]['criticality']:>4}    "
        f"{GESTURE_NAMES[gest_i]:<22} {error_rates[gest_i] * 100:>5.1f}% {tag}"
    )
    assignment_rows.append(
        {
            "command": COMMANDS[cmd_i]["name"],
            "criticality": COMMANDS[cmd_i]["criticality"],
            "gesture": GESTURE_NAMES[gest_i],
            "error_pct": round(error_rates[gest_i] * 100, 2),
            "status": tag,
        }
    )

assignment_df = pd.DataFrame(assignment_rows)
assignment_df.to_csv(os.path.join(RESULTS_DIR, "safe_assignment_table.csv"), index=False)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
sns.heatmap(
    cost_matrix,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd",
    xticklabels=GESTURE_NAMES,
    yticklabels=[c["name"] for c in COMMANDS],
    ax=ax1,
    linewidths=0.5,
)
ax1.set_title("Risk Cost Matrix")
ax1.tick_params(axis="x", rotation=45)
for cmd_i, gest_i in zip(row_ind, col_ind):
    ax1.add_patch(plt.Rectangle((gest_i, cmd_i), 1, 1, fill=False, edgecolor="blue", lw=3, clip_on=False))

bars = ax2.bar(
    ["Worst", "Naive", "Optimal"],
    [worst_cost, naive_cost, optimal_cost],
    color=["#e74c3c", "#f39c12", "#2ecc71"],
    edgecolor="black",
)
ax2.bar_label(bars, fmt="%.2f", fontsize=12, fontweight="bold")
ax2.set_ylabel("Risk Score")
ax2.set_title(f"Safety Comparison | Risk down {improvement:.1f}%", fontweight="bold")
ax2.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(RESULTS_DIR, "safe_assignment.png"), dpi=150)
plt.show()

if RUN_STREAMING_DEMO:
    print_banner("DEPLOYMENT SIMULATION (offline replay)")
    stream_rows = build_streaming_demo_rows(y_true, y_pred, y_prob, entropy, CONFIDENCE_GATE, gesture_to_command)
    pd.DataFrame(stream_rows[:500]).to_csv(
        os.path.join(RESULTS_DIR, "deployment_simulation_replay.csv"), index=False
    )
    save_json(
        os.path.join(RESULTS_DIR, "deployment_api_example.json"),
        {
            "sample_rate_hz": SR,
            "window_size_samples": WIN_SIZE,
            "channels": CHANNEL_NAMES,
            "confidence_gate": CONFIDENCE_GATE,
            "example_output": stream_rows[0],
            "simulation_note": (
                "This replay uses stored test-set windows. Live deployment "
                "requires serial/BLE ingestion upstream of this inference pipeline."
            ),
        },
    )
    save_json(
        os.path.join(RESULTS_DIR, "deployment_api_contract.json"),
        {
            "version": "1.0",
            "endpoint": "POST /predict",
            "request_schema": {
                "window": {
                    "type": "array",
                    "shape": [200, 6],
                    "dtype": "float32",
                    "channels": CHANNEL_NAMES,
                    "units": {
                        "accel": "m/s^2 (device-released or deployment-side normalized)",
                        "gyro": "rad/s (device-released or deployment-side normalized)",
                    },
                }
            },
            "response_schema": {
                "gesture": "string",
                "confidence": "float [0,1]",
                "accepted": "bool (confidence >= gate)",
                "assigned_command": "string | MANUAL_REVIEW",
                "entropy_nats": "float",
            },
            "confidence_gate": CONFIDENCE_GATE,
            "latency_budget_ms": 100,
            "simulation_note": (
                "Offline replay validated. Live deployment requires an upstream "
                "IMU driver that segments incoming sensor frames into 200-sample "
                "windows at 200 Hz and forwards them to this endpoint."
            ),
        },
    )


# -----------------------------------------------------------------------------
# 7. Creative Visualization / Dashboard
# -----------------------------------------------------------------------------
print_banner("INTERPRETABILITY, VISUALIZATION, AND DASHBOARD")

_, _, f1s, _ = precision_recall_fscore_support(y_true, y_pred, average=None)

fig, axes = plt.subplots(2, 3, figsize=(22, 12))

ax = axes[0, 0]
ax.bar(
    subjects_list,
    [a * 100 for a in accs],
    color=["#2ecc71" if a >= overall_acc else "#e74c3c" for a in accs],
    edgecolor="black",
    linewidth=0.3,
)
ax.axhline(y=overall_acc * 100, color="black", linestyle="--", label=f"Mean: {overall_acc * 100:.1f}%")
ax.set_title("Per-Subject Accuracy (LOSO)", fontweight="bold")
ax.set_ylim(0, 108)
ax.legend(fontsize=9)
ax.grid(axis="y", alpha=0.3)
ax.set_xlabel("Subject ID")
ax.set_ylabel("Accuracy (%)")

ax = axes[0, 1]
bars = ax.barh(GESTURE_NAMES, f1s * 100, color="#3498db", edgecolor="black", linewidth=0.3)
for bar, f in zip(bars, f1s):
    ax.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height() / 2, f"{f * 100:.1f}%", va="center", fontsize=9)
ax.set_title("Per-Gesture F1 Score", fontweight="bold")
ax.set_xlim(0, 108)

ax = axes[0, 2]
c_auc = ["#e74c3c" if a < 0.95 else "#2ecc71" for a in auc_scores]
ax.barh(GESTURE_NAMES, auc_scores, color=c_auc, edgecolor="black", linewidth=0.3)
ax.axvline(0.95, color="#e74c3c", ls="--", lw=2, label="0.95 threshold")
ax.axvline(mean_auc, color="black", ls="-", lw=1.5, label=f"Mean {mean_auc:.3f}")
ax.set_title("Per-Gesture ROC-AUC", fontweight="bold")
ax.set_xlim(0.85, 1.01)
ax.legend(fontsize=8)
ax.grid(axis="x", alpha=0.3)

ax = axes[1, 0]
sns.heatmap(
    cm_pct,
    annot=True,
    fmt=".0f",
    cmap="Blues",
    ax=ax,
    xticklabels=[g[:9] for g in GESTURE_NAMES],
    yticklabels=[g[:9] for g in GESTURE_NAMES],
    linewidths=0.3,
)
ax.set_title("Confusion Matrix (%)", fontweight="bold")
ax.tick_params(axis="x", rotation=45)

ax = axes[1, 1]
ax.hist(max_conf[correct], bins=50, alpha=0.6, color="#2ecc71", label="Correct", density=True)
ax.hist(max_conf[~correct], bins=50, alpha=0.6, color="#e74c3c", label="Wrong", density=True)
ax.axvline(0.7, color="black", ls="--", lw=2.5, label="Gate @ 0.7")
ax.set_xlabel("Confidence")
ax.set_ylabel("Density")
ax.set_title("Confidence Gating", fontweight="bold")
ax.legend(fontsize=8)

ax = axes[1, 2]
ax.axis("off")
best_idx = int(np.argmax(accs))
worst_idx = int(np.argmin(accs))
mean_lat = lat_mean if lat_mean is not None else 0.0
baseline_acc = baseline_summary["overall_accuracy"] * 100 if baseline_summary is not None else None

kpi_text = (
    "KPI SUMMARY\n\n"
    f"  Accuracy    {overall_acc * 100:.1f}%\n"
    f"  Weighted F1 {overall_f1 * 100:.1f}%\n"
    f"  Mean AUC    {mean_auc:.4f}\n"
    f"  Std (subj)  {np.std(accs) * 100:.1f}%\n\n"
    f"  Best  subj  #{all_results[best_idx]['test_subject']:02d} ({max(accs) * 100:.1f}%)\n"
    f"  Worst subj  #{all_results[worst_idx]['test_subject']:02d} ({min(accs) * 100:.1f}%)\n\n"
    "DEPLOYMENT\n\n"
    f"  Latency     {mean_lat:.1f} ms\n"
    "  Budget      100 ms\n"
    f"  Status      {'PASS' if mean_lat < 100 else 'FAIL'}\n\n"
    "SAFETY\n\n"
    f"  Risk down   {improvement:.1f}% vs naive\n"
    "  Assignment  Hungarian\n"
    f"  Gate@{CONFIDENCE_GATE:.2f}   active\n\n"
    "MODEL\n\n"
    "  Arch: CNN-LSTM\n"
    f"  Input: ({WIN_SIZE}, 6) IMU\n"
    f"  Preproc: {PREPROCESS_MODE}\n"
    f"  SR: {SR} Hz"
)
if baseline_acc is not None:
    kpi_text += f"\n  Baseline    {baseline_acc:.1f}%"

ax.text(
    0.05,
    0.97,
    kpi_text,
    transform=ax.transAxes,
    fontsize=10,
    va="top",
    fontfamily="monospace",
    bbox=dict(boxstyle="round,pad=0.6", facecolor="white", edgecolor="#e74c3c", linewidth=2.5),
)

plt.suptitle("HGAG Gesture Recognition -- Executive Dashboard (LOSO)", fontsize=15, fontweight="bold", y=1.01)
plt.tight_layout()
plt.savefig(os.path.join(RESULTS_DIR, "dashboard.png"), dpi=150, bbox_inches="tight")
plt.show()


# -----------------------------------------------------------------------------
# 8. Final documentation exports
# -----------------------------------------------------------------------------
print_banner("FINAL DOCUMENTATION EXPORTS")

final_summary = {
    "overall_accuracy": round(float(overall_acc), 6),
    "overall_accuracy_pct": round(float(overall_acc * 100), 2),
    "weighted_f1": round(float(overall_f1), 6),
    "mean_auc": round(mean_auc, 6),
    "recommended_gate": CONFIDENCE_GATE,
    "gated_accuracy_at_0_70_pct": round(float(gating_df.loc[np.isclose(gating_df["threshold"], CONFIDENCE_GATE), "gated_acc"].iloc[0]), 2),
    "accepted_at_0_70_pct": round(float(gating_df.loc[np.isclose(gating_df["threshold"], CONFIDENCE_GATE), "accept_pct"].iloc[0]), 2),
    "risk_reduction_pct": round(improvement, 2),
    "latency_mean_ms": round(lat_mean, 2) if lat_mean is not None else None,
    "latency_max_ms": round(lat_max, 2) if lat_max is not None else None,
    "latency_budget_pass_folds": lat_budget_pass,
    "n_folds": len(all_results),
    "n_windows": int(len(y_true)),
    "preprocess_mode": PREPROCESS_MODE,
    "baseline_accuracy_pct": round(baseline_summary["overall_accuracy"] * 100, 2) if baseline_summary is not None else None,
    "calibration_ece": round(float(ece), 6),
    "calibration_mce": round(float(mce), 6),
    "class_balance_cv": round(float(class_balance_summary["coefficient_of_variation"]), 6),
    "class_balance_gini": round(float(class_balance_summary["gini_index"]), 6),
    "top_confused_pairs": [
        {"count": int(cnt), "true_gesture": g_true, "pred_gesture": g_pred}
        for cnt, g_true, g_pred in pairs[:5]
    ],
}
save_json(os.path.join(RESULTS_DIR, "final_metrics_summary.json"), final_summary)
save_final_project_summary(RESULTS_DIR, final_summary)

rubric_coverage = {
    "problem_understanding_and_dataset_analysis": {
        "implemented": True,
        "artifacts": [
            "runtime_audit_summary.json",
            "eda_dataset_overview.png",
            "eda_correlation.png",
            "eda_profiles.png",
            "eda_energy.png",
            "eda_released_signal_example.png",
        ],
    },
    "model_design_and_justification": {
        "implemented": True,
        "artifacts": [
            "model_summary.txt",
            "architecture_rationale.txt",
            "benchmark_comparison.csv",
            "benchmark_comparison.png",
        ],
    },
    "data_preprocessing_and_feature_engineering": {
        "implemented": True,
        "artifacts": [
            "project_run_metadata.json",
            "augmentation_preview.png",
            "baseline_feature_loso_results.csv",
            "class_balance_report.csv",
        ],
    },
    "model_training_and_evaluation": {
        "implemented": True,
        "artifacts": [
            "loso_results.json",
            "predictions.npz",
            "training_curves_all_folds.png",
            "confusion_matrix.png",
            "roc_auc_per_gesture.png",
        ],
    },
    "interpretability_and_visualization": {
        "implemented": True,
        "artifacts": [
            "confidence_entropy.png",
            "calibration_reliability.png",
            "dashboard.png",
            "xai_gradcam_examples.png" if RUN_XAI else None,
            "xai_gradcam_crossfold.png" if RUN_XAI else None,
        ],
    },
    "deployment_readiness_application_insight": {
        "implemented": True,
        "artifacts": [
            "gating_table.csv",
            "inference_latency.png",
            "safe_assignment.png",
            "safe_assignment_table.csv",
            "deployment_simulation_replay.csv" if RUN_STREAMING_DEMO else None,
            "deployment_api_contract.json" if RUN_STREAMING_DEMO else None,
        ],
    },
    "bonus_explainable_ai": {
        "implemented": RUN_XAI,
        "artifacts": [
            "xai_gradcam_examples.png",
            "xai_gradcam_examples.csv",
            "xai_gradcam_crossfold.png",
        ] if RUN_XAI else [],
    },
    "bonus_model_optimization": {
        "implemented": RUN_MODEL_OPTIMIZATION,
        "artifacts": [
            "quantization_summary.json",
            "representative_fold_model_dynamic.tflite",
            "representative_fold_model_int8.tflite",
        ] if RUN_MODEL_OPTIMIZATION else [],
    },
    "bonus_creative_dashboard": {
        "implemented": True,
        "artifacts": ["dashboard.png"],
    },
    "documentation_and_presentation": {
        "implemented": True,
        "artifacts": [
            "FINAL_PROJECT_SUMMARY.md",
            "runtime_audit_summary.json",
            "rubric_coverage_summary.json",
            "final_metrics_summary.json",
        ],
    },
}
save_rubric_coverage(RESULTS_DIR, rubric_coverage)


print(f"\n{'=' * 78}")
print("  FINAL NOTEBOOK COMPLETE")
print(f"  Results -> {RESULTS_DIR}")
print(f"{'=' * 78}")
for f in sorted(os.listdir(RESULTS_DIR)):
    sz = os.path.getsize(os.path.join(RESULTS_DIR, f)) // 1024
    icon = "PNG" if f.endswith(".png") else ("CSV" if f.endswith(".csv") else "FILE")
    print(f"  {icon}  {f:<52} {sz:>5} KB")
```
